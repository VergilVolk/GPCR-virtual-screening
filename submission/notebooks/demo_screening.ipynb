```python
import sys, numpy as np, torch
from pathlib import Path
ROOT = Path('').resolve().parent if Path('').resolve().name == 'notebooks' else Path('').resolve()
sys.path.insert(0, str(ROOT))
from src.proj import Proj
from src.screen_core import screen, write_results_csv, TRACK
```

```python
# 1) 加载示例数据（50 分子 x 10 个 GaMD M4 口袋表征）
models = sorted((ROOT/'models').glob('external_result.seed*.projection.pt'))
print('权重:', [m.name for m in models])
```

```python
# 2) 打分：每种子 10 口袋 mean 池化，3 种子平均（head-to-head 胜出配置）
ids, scores, per_seed = screen(ROOT/'data/demo/demo_molecules.npz',
                               ROOT/'data/demo/demo_m4_pockets.npz', models, 'mean')
print('seed 间 std（前10名）: %.4f' % per_seed.std(axis=0)[np.argsort(-scores)[:10]].mean())
```

```python
# 3) 输出标准化候选清单（附件5 第四节字段）
out = write_results_csv(ROOT/'results/results_nb.csv', ids, scores,
                        'DrugCLIP(2023)+triplet投影 v01', top_n=10)
print(open(out, encoding='utf-8').read())
```
